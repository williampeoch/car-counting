```python
import matplotlib.pyplot as plt
import numpy as np
import os
import cv2
from natsort import natsorted
```

```python
path="data"

im_list=[]

for filename in natsorted(os.listdir(path)):
    if filename.lower().endswith(('.jpg')):
        im = plt.imread(os.path.join(path, filename))
        im_list.append(im)

data = np.array(im_list)
data.shape
```

```python
background = np.mean(data, axis=0).astype(np.uint8)

plt.figure(figsize=(12, 6))
plt.imshow(background)
plt.axis("off")
plt.show()
```

```python
diff = cv2.absdiff(data[4], background)

plt.figure(figsize=(12, 6))
plt.imshow(diff)
plt.axis("off")
plt.show()
```
